#### Importing Libraries

```python
# Importing the required libraries

import pandas as pd
import numpy as np
import plotly.express as px
import plotly.graph_objects as go
```

#### Loading the Dataset

```python
df = pd.read_csv("../data/raw/defects_data.csv")
```

#### Univariate Analysis

#### Numerical Variable Distributions

```python
#### Distribution of Repair Cost

fig = px.histogram(
    df,
    x="repair_cost",
    nbins=30,
    title="Distribution of Repair Cost",
    labels={
        "repair_cost": "Repair Cost",
        "count": "Number of Defects"
    },
    color_discrete_sequence=["#3366CC"],
    template="plotly_white"
)

fig.update_layout(
    title_x=0.5,
    width=1200,
    height=600,
    showlegend=False,
    xaxis_title="Repair Cost",
    yaxis_title="Number of Defects"
)

fig.show()
```

#### Categorical Variable Distributions

```python
#### Distribution of Categorical Variables

categorical_columns = [
    "defect_type",
    "defect_location",
    "severity",
    "inspection_method"
]

for column in categorical_columns:

    category_counts = (
        df[column]
        .value_counts()
        .reset_index()
    )

    category_counts.columns = [column, "Count"]

    fig = px.bar(
        category_counts,
        x=column,
        y="Count",
        title=f"Distribution of {column.replace('_', ' ').title()}",
        labels={
            column: column.replace("_", " ").title(),
            "Count": "Number of Defects"
        },
        color=column,
        text="Count",
        template="plotly_white"
    )

    fig.update_traces(
        textposition="outside",
        showlegend=False
    )

    fig.update_layout(
        title_x=0.5,
        width=1200,
        height=600,
        showlegend=False,
        xaxis_title=column.replace("_", " ").title(),
        yaxis_title="Number of Defects"
    )

    fig.show()
```

#### Outlier Analysis

```python
#### Identifying Outliers Using the IQR Method

Q1 = df["repair_cost"].quantile(0.25)
Q3 = df["repair_cost"].quantile(0.75)

IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

outliers = df[
    (df["repair_cost"] < lower_bound) |
    (df["repair_cost"] > upper_bound)
]

outlier_summary = pd.DataFrame({
    "Metric": [
        "First Quartile (Q1)",
        "Third Quartile (Q3)",
        "Interquartile Range (IQR)",
        "Lower Bound",
        "Upper Bound",
        "Total Records",
        "Potential Outliers",
        "Outlier Percentage"
    ],
    "Value": [
        Q1,
        Q3,
        IQR,
        lower_bound,
        upper_bound,
        len(df),
        len(outliers),
        (len(outliers) / len(df)) * 100
    ]
})

outlier_summary
```

```python
#### Box Plot of Repair Cost

fig = px.box(
    df,
    y="repair_cost",
    title="Outlier Analysis of Repair Cost",
    labels={
        "repair_cost": "Repair Cost"
    },
    color_discrete_sequence=["#3366CC"],
    template="plotly_white"
)

fig.update_layout(
    title_x=0.5,
    width=1200,
    height=600,
    showlegend=False,
    yaxis_title="Repair Cost"
)

fig.show()
```

#### Observations on Outlier Analysis

- The IQR method identified **0 potential outliers (0%)** in `repair_cost` across the 1,000 records.
- The lower and upper IQR bounds were **−461.34** and **1,491.31**, respectively. All observed repair costs fall within these boundaries.
- The lower bound is negative, while repair costs are non-negative, indicating that the IQR boundaries are relatively broad compared with the observed cost range.
- No records were flagged for outlier treatment based on the IQR method. However, this does not establish that every repair cost is typical or free from unusual business cases.
